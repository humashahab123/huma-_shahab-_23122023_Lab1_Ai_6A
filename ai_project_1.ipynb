***TRAVEL AGENT CODE***

```python
flights = [
    {"id": "F101", "from": "Karachi", "to": "Lahore", "price": 12000},
    {"id": "F102", "from": "Karachi", "to": "Islamabad", "price": 15000},
    {"id": "F103", "from": "Lahore", "to": "Islamabad", "price": 8000},
    {"id": "F104", "from": "Karachi", "to": "Dubai", "price": 45000},
    {"id": "F105", "from": "Islamabad", "to": "Dubai", "price": 42000},
]


def search_flights(departure, destination):
    matches = []
    for flight in flights:
        if flight["from"].lower() == departure.lower() and flight["to"].lower() == destination.lower():
            matches.append(flight)
    return matches


def display_flights(matches):
    print("\nAvailable flights:")
    for f in matches:
        print(f"  ID: {f['id']} | {f['from']} -> {f['to']} | Price: Rs {f['price']}")


def book_flight(matches, chosen_id):
    for f in matches:
        if f["id"].lower() == chosen_id.lower():
            return f
    return None


def travel_agent():
    print("START")

    departure = input("Enter departure city: ").strip()
    destination = input("Enter destination city: ").strip()

    matches = search_flights(departure, destination)

    if len(matches) == 0:
        print("No flights available")
        print("STOP")
        return

    else:
        display_flights(matches)

        chosen_id = input("\nAsk user to choose a flight ID: ").strip()
        booked = book_flight(matches, chosen_id)

        if booked is None:
            print("Invalid flight ID selected. Booking cancelled.")
        else:
            print(f"\nFlight booked successfully! ({booked['id']}: {booked['from']} -> {booked['to']}, Rs {booked['price']})")

    print("END")


travel_agent()
```
